# Create logging tables

Run by hand, once per environment. Creates `` `finhive-2026`.logs `` and every logging table the pipelines write to.

```python
spark.sql("CREATE CATALOG IF NOT EXISTS `finhive-2026`")
spark.sql("CREATE SCHEMA IF NOT EXISTS `finhive-2026`.logs")
```

```python
ingestion_log_schema = "series string, source string, status boolean, updateAt timestamp, lastObservationDate date, item_count int, error string, job_name string"

if not spark.catalog.tableExists("`finhive-2026`.logs.ingestionLog"):
    spark.createDataFrame([], ingestion_log_schema).write.saveAsTable("`finhive-2026`.logs.ingestionLog")
```
