# Create logging tables

Run by hand, once per environment. Creates `finhive.logs` and every logging table the pipelines write to.

```python
spark.sql("CREATE CATALOG IF NOT EXISTS finhive")
spark.sql("CREATE SCHEMA IF NOT EXISTS finhive.logs")
```

```python
ingestion_log_schema = "series string, source string, status boolean, updateAt timestamp, lastObservationDate date, item_count int, error string, job_name string"

if not spark.catalog.tableExists("finhive.logs.ingestionLog"):
    spark.createDataFrame([], ingestion_log_schema).write.saveAsTable("finhive.logs.ingestionLog")
```
